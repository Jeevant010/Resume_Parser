## 1. Text Extraction
We extract raw text from a sample PDF resume.

```python
import fitz  # PyMuPDF
import os

pdf_path = '../data/Alice Clark CV.pdf'
text = ''
if os.path.exists(pdf_path):
    with fitz.open(pdf_path) as doc:
        for page in doc:
            text += page.get_text() + ' '
    text = ' '.join(text.split())
    print('Extracted Text length:', len(text))
    print(text[:500], '...')
else:
    print('Sample PDF not found.')
```

## 2. Zero-Shot Entity Extraction
We use GLiNER to extract entities without training.

```python
from gliner import GLiNER

if text:
    # Initialize GLiNER with the standard model
    model = GLiNER.from_pretrained('urchade/gliner_medium-v2.1')

    # Define labels we want to extract
    labels = ['Name', 'Email', 'Phone', 'Location', 'Experience', 'Education', 'Skills']

    # Predict
    entities = model.predict_entities(text, labels)

    # Display formatted
    for entity in entities:
        print(f"{entity['label']}: {entity['text']}")
```
